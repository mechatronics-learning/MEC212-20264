# 2.6 Input/Output Statement

## Input Statement: `input()`

The `input()` function is used to receive input from the keyboard. The program pauses until the user enters a value and presses **Enter**.

```python
name = input("Please enter your name: ")
print(f"Hello, {name}!")

age = int(input("Enter your age: "))
print(f"You will be {age + 1} next year.")

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
print(f"The sum is: {num1 + num2}")
```

## Output Statement: `print()`

The `print()` function is used to display output on the screen.

```python
print("Welcome to Python Programming!")

score = 95
name = "Charlie"
print("Student:", name, "Score:", score)
print(f"Student: {name}, Score: {score}")
print("Apple", "Banana", "Cherry", sep=" | ")
print("Loading...", end=" ")
print("Done!")
print()
```
