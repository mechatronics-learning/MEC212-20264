## Input Statement: `input()`

The `input()` function receives information from the keyboard. The program waits until the user enters a value and presses **Enter**.

````python
speed = input("Enter motor speed (rpm): ")
print("Motor Speed =", speed, "rpm")
````

### Input with numerical values

Use `int()` for whole numbers and `float()` for decimal numbers.

````python
age = int(input("Enter your age: "))
print("Next year, your age will be", age + 1)

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
total = num1 + num2
print("Total =", total)
````

## Output Statement: `print()`

The `print()` function displays messages, variable values and calculation results.

````python
print("Welcome to Python Programming!")

temperature = 75
speed = 1500
print("Temperature =", temperature, "°C")
print("Motor Speed =", speed, "rpm")
````
