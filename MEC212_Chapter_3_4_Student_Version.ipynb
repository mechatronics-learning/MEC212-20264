### Quick notes
| Structure | Purpose |
|---|---|
| `if` | Run an action only when a condition is True |
| `if...else` | Choose one of two actions |
| `if...elif...else` | Check conditions from top to bottom; execute the first True branch |
| `match...case` | Select based on a matching value; `case _` is the default |

Relational operators: `>`, `<`, `>=`, `<=`, `==`, `!=`. Logical operators: `and`, `or`, `not`.


## Worked Examples (1–4)

Study these four examples before attempting Questions 1–20. Each example shows the problem followed by a complete Python solution.


### Example 1: One-way Selection (`if`)

**Problem:** A student has an attendance percentage of 92%. Display `Good attendance` only if attendance is at least 80%. Always display `Attendance checked`.

**Answer:**


```python
attendance = 92
if attendance >= 80:
    print("Good attendance")
print("Attendance checked")
```

### Example 2: Two-way Selection (`if...else`)

**Problem:** A motor temperature is 75°C. Display `Temperature normal` if the temperature is 80°C or below. Otherwise, display `Temperature high`.

**Answer:**


```python
temperature = 75
if temperature <= 80:
    print("Temperature normal")
else:
    print("Temperature high")
```

### Example 3: Multi-way Selection (`if...elif...else`)

**Problem:** A student scores 72 marks. Display `Excellent` for marks of 80 or above, `Good` for marks of 60 or above, and `Needs improvement` otherwise.

**Answer:**


```python
mark = 72
if mark >= 80:
    print("Excellent")
elif mark >= 60:
    print("Good")
else:
    print("Needs improvement")
```

### Example 4: Multi-way Selection (`match...case`)

**Problem:** A workshop machine has mode 2 selected. Use `match...case` to display `Manual` for mode 1, `Automatic` for mode 2, `Maintenance` for mode 3, and `Invalid mode` otherwise.

**Answer:**


```python
mode = 2
match mode:
    case 1:
        print("Manual")
    case 2:
        print("Automatic")
    case 3:
        print("Maintenance")
    case _:
        print("Invalid mode")
```

### Quick review
- **`if`:** one-way decision.
