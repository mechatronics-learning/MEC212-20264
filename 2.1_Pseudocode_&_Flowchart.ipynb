**Exercise 2.1 – Pseudocode**

Question - Write a pseudocode to calculate the area of a rectangle.

The program should:

Read the length.
Read the width.
Calculate the area.
Display the area.
Task

Write the complete pseudocode.

```python
# BEGIN
#     INPUT length
#     INPUT width
#     SET area ← length * width
#     DISPLAY area
# END
```

```python
#Translating Pseudocode into Python
length = float(input("Enter length: "))
width = float(input("Enter width: "))
area = length * width
print("Area =", area)
```
