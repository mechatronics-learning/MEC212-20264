# 5.2 Introduction to Programmer-Defined Functions

A **programmer-defined function** is a function created by the programmer to perform a specific task. It is defined using the `def` keyword.

### General Syntax

```python
def function_name(parameters):
    # Function body
    return value
```

### Example

```python
def calculate_area(length, width):
    area = length * width
    return area
```

| Part | Description |
|---|---|
| `def` | Keyword used to define a function |
| `function_name` | Meaningful name that follows Python identifier rules |
| `parameters` | Optional inputs received by the function |
| `:` | Indicates the start of the function body |
| Indented statements | Instructions performed by the function |
| `return` | Sends a value back to the calling program (optional) |

**Notes**
- A function may have zero or more parameters.
- A function may return a value or not return any value.
- The function body must be indented.

## Concepts: Parameters, Arguments and Return Values

A **parameter** is a variable declared in the function definition.

```python
def calculate_force(mass, acceleration):
    return mass * acceleration
```

Here, `mass` and `acceleration` are parameters.

An **argument** is the actual value passed when the function is called.

```python
force = calculate_force(10, 9.81)
```

Here, `10` and `9.81` are arguments.

A **return value** is the value sent back to the caller.

````python
def calculate_force(mass, acceleration):
    return mass * acceleration

force = calculate_force(10, 9.81)
print(force)
````

> **Note:** The returned value for `calculate_force(10, 9.81)` is **98.1**, not 9.81.

## Function Definition – Engineering Example

Calculate strain given stress and elastic modulus:

**Strain = Stress / E**

Stress is given in MPa and Young's modulus is given in GPa, so the modulus is converted to MPa by multiplying by 1000.

````python
def calculate_strain(stress, elastic_modulus):
    """
    Calculate strain from stress and elastic modulus.

    Parameters:
        stress (float): Stress in MPa
        elastic_modulus (float): Young's modulus in GPa

    Returns:
        float: Strain (dimensionless)
    """
    strain = stress / (elastic_modulus * 1000)
    return strain

strain = calculate_strain(200, 200)
print("Strain =", strain)
````

## Calling a Function

A function must be defined before it is called.

````python
# Step 1: Define the function
def greet(name):
    return f"Hello, {name}!"

# Step 2: Call the function
message = greet("Alice")
print(message)

# Step 3: Call it again (reuse)
message2 = greet("Bob")
print(message2)
````

### Execution Flow

1. `greet("Alice")` is encountered.
2. Control transfers into the function; `name = "Alice"`.
3. Return value is computed: `"Hello, Alice!"`.
4. Control returns to the caller; `message` receives the return value.
5. Execution continues with `print(message)`.
