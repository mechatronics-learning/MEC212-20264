## 2.2 Identifiers, Variables, Constants & Reversed Words

Identifiers are names for data in your program.Examples:
| Term                        | Description                                                                                  | Example                                         |
| --------------------------- | -------------------------------------------------------------------------------------------- | ----------------------------------------------- |
| **Identifier**              | A name given to a variable, function, class, or other object.                                | `student_name`, `calculate_area`                |
| **Variable**                | A named storage location whose value can change during program execution.                    | `age = 20`                                      |
| **Constant**                | A value intended not to change. In Python, constants are written in uppercase by convention. | `PI = 3.14159`, `GRAVITY = 9.81`                |
| **Reserved Word (Keyword)** | A word reserved by Python with a special meaning. It cannot be used as an identifier.        | `if`, `for`, `while`, `def`, `import`, `return` |


````python
# Example 1: Clear engineering variable names
shaft_diameter_mm = 32

print(shaft_diameter_mm)
````

````python
# Example 2: Constants by convention (ALL_CAPS)
GRAVITY = 9.81
STEEL_DENSITY = 7850

print(f"g = {GRAVITY} m/s^2, rho_steel = {STEEL_DENSITY} kg/m^3")
````

````python
# Example 3: Variables can change during process monitoring
tool_temperature_c = 45
print(f"Initial tool temperature: {tool_temperature_c} degC")

tool_temperature_c = 52
print(f"Updated tool temperature: {tool_temperature_c} degC")
````

## 2.3 Basic Data Types in Python
Data types define the kind of value stored: integer, decimal, text, or true/false.
Mechanical example: part count (int), diameter (float), sensor label (str), safety flag (bool).

````python
# Example 1: int and float
bolt_count = 12
shaft_diameter_mm = 19.5

print(type(bolt_count), bolt_count)
print(type(shaft_diameter_mm), shaft_diameter_mm)
````

````python
#  Example 2: strings for labels and reports
sensor_id = "TC-07"
unit = "degC"
message = f"Sensor {sensor_id} reading unit: {unit}"
print(message)
````

````python
#  Example 3: bool from a safety comparison
stress_mpa = 180
yield_strength_mpa = 250
is_safe = stress_mpa < yield_strength_mpa

print(f"Safe condition? {is_safe}")
````
