## What is a Function?
A **function** is a reusable block of code that performs a specific task. Functions improve readability, maintainability, testing, reuse, and debugging.

| Built-in functions | Programmer-defined functions |
|---|---|
| Provided by Python | Written by the programmer |
| `print()`, `len()`, `input()` | `calculate_area()`, `calculate_stress()` |

## Function Pseudocode
**Terminology:** `FUNCTION` begins a definition; parameters receive values; `LOCAL` declares an optional local variable; `SET` assigns a value; `RETURN` sends a result to the caller; `END FUNCTION` ends the definition.

```text
FUNCTION function_name(parameter1, parameter2)
    LOCAL result
    SET result = calculation
    RETURN result
END FUNCTION
```

**Example – Celsius to Fahrenheit:**
```text
FUNCTION celsius_to_fahrenheit(celsius)
    LOCAL fahrenheit
    SET fahrenheit = (celsius * 9/5) + 32
    RETURN fahrenheit
END FUNCTION

SET temp_c = 25
SET temp_f = celsius_to_fahrenheit(temp_c)
PRINT temp_f
```

````python
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

temp_c = 25
temp_f = celsius_to_fahrenheit(temp_c)
print(temp_f)  # 77.0
````

## Flowchart for a Function
A function flowchart shows **entry/parameters → processing → return → end**. Use a terminal for START/END and a rectangle for calculation. A return can be shown as a labelled output/return step. A function parameter is received from the caller; it is not necessarily read with `input()`.

```text
 (START: celsius_to_fahrenheit(celsius))
                  |
                  v
 [fahrenheit = (celsius * 9/5) + 32]
                  |
                  v
        / RETURN fahrenheit /
                  |
                  v
               (END)
```

## Real Engineering Example – Axial Stress
Stress: **σ = F / A**. When force is in **N** and area in **m²**, stress is in **Pa**. Assume `area > 0`.

**Without functions:**

````python
force = 500
area = 0.05
stress = force / area
print("Case 1:", stress, "Pa")

force = 800
area = 0.08
stress = force / area
print("Case 2:", stress, "Pa")
````

**With a programmer-defined function:**

````python
def calculate_stress(force, area):
    return force / area

print("Case 1:", calculate_stress(500, 0.05), "Pa")
print("Case 2:", calculate_stress(800, 0.08), "Pa")
# Both results: 10000.0 Pa (0.01 MPa)
````
