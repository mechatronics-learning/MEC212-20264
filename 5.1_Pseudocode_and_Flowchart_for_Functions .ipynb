## What is a Function?

A **function** is a reusable block of code that performs a specific task.

Instead of writing the same code repeatedly, we place it inside a function and call the function whenever needed.

Functions make programs:
- easier to read,
- easier to maintain,
- easier to test,
- easier to reuse,
- easier to debug.

### Built-in vs Programmer-Defined Functions

| Built-in Function | Programmer-Defined Function |
|---|---|
| Already provided by Python | Created by the programmer |
| Examples: `print()`, `len()`, `input()` | Examples: `calculate_area()`, `calculate_bmi()` |

## Pseudocode for Functions

Function pseudocode shows the interface (**name, inputs, outputs**) and the algorithm (**steps inside**) using plain language.

### General Form

```text
FUNCTION function_name(parameter1, parameter2)
    LOCAL variable declarations
    algorithm steps
    RETURN result
END FUNCTION
```

### Example — Convert Celsius to Fahrenheit

```text
FUNCTION celsius_to_fahrenheit(celsius)
    LOCAL fahrenheit
    SET fahrenheit = (celsius × 9/5) + 32
    RETURN fahrenheit
END FUNCTION

INPUT temp_c
SET temp_f = celsius_to_fahrenheit(temp_c)
PRINT temp_f
```

### Python Equivalent

````python
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

temp_c = 25
temp_f = celsius_to_fahrenheit(temp_c)
print(temp_f)
````

## Flowchart for Functions

A flowchart for a function shows:
- **Input parameters**
- **Processing steps**
- **Return value**
- **Back to caller**

Example:

```text
Start: celsius_to_fahrenheit(celsius)
                ↓
fahrenheit = (celsius × 9/5) + 32
                ↓
Return fahrenheit
                ↓
Back to caller
```

## Real Engineering Example

### Without Functions

The same calculation is repeated.

````python
force = 500
area = 0.05
stress = force / area
print(stress)

force = 800
area = 0.08
stress = force / area
print(stress)
````

### With Functions

Functions reduce repeated code and improve readability.

````python
def calculate_stress(force, area):
    return force / area

print(calculate_stress(500, 0.05))
print(calculate_stress(800, 0.08))
````
