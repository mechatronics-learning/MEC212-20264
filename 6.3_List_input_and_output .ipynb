## 6.3 Input and Output of Arrays

**Output** — print all elements in a loop, with optional formatting using `enumerate()` to get both the index and value:

```python
for i, temp in enumerate(temperatures):
    print(f"Sensor {i}: {temp:.1f}°C")
```

**Input** — build an array by appending user-supplied values one by one:

```python
readings = []
for i in range(n):
    val = float(input(f"Enter reading {i+1}: "))
    readings.append(val)
```

In Google Colab, `input()` creates an interactive text box. For auto-runnable solutions, hardcoded values are used instead.

## 6.4 Operations on Arrays: Sum, Average, Min, Max

Four fundamental statistical operations that Python provides as built-in functions:

| Operation | Built-in | Manual loop approach |
|---|---|---|
| **Sum** | `sum(arr)` | Accumulate with `total += x` |
| **Average** | `sum(arr) / len(arr)` | `total / count` after loop |
| **Minimum** | `min(arr)` | Track `min_val`, update if smaller |
| **Maximum** | `max(arr)` | Track `max_val`, update if larger |

Always verify results make physical sense — units, realistic ranges, expected order of magnitude.


````python
# Worked example — full analysis of beam deflection data
deflections_mm = [0.0, 0.5, 1.2, 2.1, 3.0, 3.5, 2.8, 1.5, 0.8, 0.0]

total_defl = sum(deflections_mm)
avg_defl   = total_defl / len(deflections_mm)
max_defl   = max(deflections_mm)
min_defl   = min(deflections_mm)
peak_idx   = deflections_mm.index(max_defl)

print(f"Number of measurement points: {len(deflections_mm)}")
print(f"Sum of deflections : {total_defl:.1f} mm")
print(f"Average deflection : {avg_defl:.2f} mm")
print(f"Maximum deflection : {max_defl:.1f} mm at point {peak_idx}")
print(f"Minimum deflection : {min_defl:.1f} mm")
print(f"Deflection range   : {max_defl - min_defl:.1f} mm")
````
