## 4.2.1 `for` Loop — Count-Controlled (Definite) Repetition

**Learning outcomes**
- Explain why a `for` loop is used for repeated tasks.
- Write a `for` loop with correct colon (`:`) and indentation.
- Use `range(stop)`, `range(start, stop)` and `range(start, stop, step)`.
- Repeat calculations, collect input and process a list using a `for` loop.

**Recall:** A repetition control structure runs a block of statements repeatedly. A `for` loop is commonly used when the number of repetitions is known or when processing each value in a sequence.

**Important:** The stopping value of `range()` is **excluded**.

### How to write a `for` loop
```python
for i in range(5):
    print(i)
```
- `for` is the Python keyword.
- `i` is the loop variable used throughout this notebook.
- `range(5)` generates 0, 1, 2, 3, 4.
- `:` is required at the end of the `for` line.
- The indented statement is repeated.
- Code without indentation after the loop runs after the repetition ends.

### Understanding `range()`
| Expression | Values generated |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(2, 7)` | 2, 3, 4, 5, 6 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

`range(start, stop, step)` allows a starting value and a step. The stop value is never included.

Run these examples before attempting the questions.

### Example 1: Display Numbers from 1 to 5

````python
for i in range(1, 6):
    print(i)
````

### Example 2: Calculate the Total Cost of Five Items

````python
total = 0
for i in range(5):
    price = float(input("Enter item price (RM): "))
    total = total + price
print("Total cost: RM", round(total, 2))
````

### Example 3: Display Student Marks from a List

````python
marks = [70, 85, 90, 65, 80]
for i in marks:
    print("Student mark:", i)
````

### Example 4: Display a Multiplication Table

````python
for i in range(1, 6):
    result = 3 * i
    print("3 x", i, "=", result)
````

### Example 5: Calculate Total and Average Force

````python
forces = [120.5, 115.2, 130.8, 118.4, 125.0]
total = 0
for i in forces:
    total = total + i
average = total / len(forces)
print("Total force:", round(total, 2), "N")
print("Average force:", round(average, 2), "N")
````

### Example 6: Print Even Numbers

````python
for i in range(2, 11, 2):
    print(i)
````

### Example 7: Total and Average Torque (Engineering)
Five torque readings are `[120, 135, 118, 142, 130]` N·m. Use a `for` loop to find the total and average.


````python
torques = [120, 135, 118, 142, 130]
total = 0
for i in range(5):
    total = total + torques[i]
average = total / 5
print("Total torque (N.m):", total)
print("Average torque (N.m):", round(average, 2))
````
