## 2.5 Assignment Statement
An **assignment statement** stores a value in a variable. The `=` operator assigns the value on the right to the variable on the left.

`variable_name = value`

### Example 1 – Assign a value

````python
score = 10
print(score)
````

### Example 2 – Reassign a value

````python
score = 10
score = 15
print(score)
````

### Example 3 – Update using an expression

````python
score = 15
score = score + 5
print(score)
````

## 2.5.1 Basic Assignment Statements
Variables can store strings, integers, floats and Booleans.

````python
student_name = "Aina"
student_age = 19
cgpa = 3.75
is_registered = True
print("Student Name =", student_name)
print("Student Age =", student_age)
print("CGPA =", cgpa)
print("Registration Status =", is_registered)
````

## 2.5.2 Reassignment and Mathematical Expressions
A new assignment can replace an old value, or store the result of an expression.

````python
tool_temperature = 45
print("Initial Temperature =", tool_temperature, "°C")
tool_temperature = 52
print("Updated Temperature =", tool_temperature, "°C")
````

````python
length = 8.5
width = 4.0
area = length * width
perimeter = 2 * (length + width)
print("Area =", area, "m²")
print("Perimeter =", perimeter, "m")
````

## 2.5.3 Augmented Assignment Operators
| Operator | Example | Equivalent statement |
|---|---|---|
| `+=` | `x += 5` | `x = x + 5` |
| `-=` | `x -= 2` | `x = x - 2` |
| `*=` | `x *= 3` | `x = x * 3` |
| `/=` | `x /= 2` | `x = x / 2` |
| `//=` | `x //= 2` | `x = x // 2` |
| `%=` | `x %= 3` | `x = x % 3` |
| `**=` | `x **= 2` | `x = x ** 2` |

````python
score = 10
score += 5
print("After Addition =", score)
score -= 3
print("After Subtraction =", score)
score *= 2
print("After Multiplication =", score)
score /= 4
print("After Division =", score)
````

## 2.5.4 Multiple Assignment
Python allows multiple variables to receive values in one statement.

````python
length, width, height = 5, 3, 2
volume = length * width * height
print("Volume =", volume, "m³")
````

````python
sensor_1 = sensor_2 = sensor_3 = 0
print("Sensor 1 =", sensor_1)
print("Sensor 2 =", sensor_2)
print("Sensor 3 =", sensor_3)
````

````python
motor_a_speed = 1200
motor_b_speed = 1500
print("Before Swapping:")
print("Motor A =", motor_a_speed, "rpm")
print("Motor B =", motor_b_speed, "rpm")
motor_a_speed, motor_b_speed = motor_b_speed, motor_a_speed
print("After Swapping:")
print("Motor A =", motor_a_speed, "rpm")
print("Motor B =", motor_b_speed, "rpm")
````
