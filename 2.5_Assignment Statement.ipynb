# 2.5 Assignment Statement

In Python, an assignment statement is a fundamental operation used to bind a value to a variable name. It's how you store data in your program's memory so you can refer to and manipulate it later.
The most common form of an assignment statement uses the assignment operator (=)

```python
variable_name = value
```



````python
# Example 1: Set the variable value to 10
score = 10
print(score)
````

````python
# Example 2: Update the variable value to 15
score = 15
print(score)
````

````python
# Example 3: Assign the result back to the same variable
score = score + 5
print(score)
````

## 2.5.1 Basic Assignment Statements

An assignment statement stores a value in a variable using the assignment operator `=`.

```python
variable_name = value
```

The expression on the right-hand side is evaluated first, and the result is assigned to the variable on the left-hand side.

````python
# Example 4: Assign different data types to variables

student_name = "Aina"
student_age = 19
cgpa = 3.75
is_registered = True

print("Student Name:", student_name)
print("Student Age:", student_age)
print("CGPA:", cgpa)
print("Registration Status:", is_registered)
````

## 2.5.2 Intermediate Assignment Statements

A variable can be reassigned, and the new value replaces its previous value.

An assignment statement may also use an arithmetic expression on the right-hand side.

````python
# Intermediate Example 1: Reassignment

tool_temperature = 45
print("Initial tool temperature:", tool_temperature, "degC")

tool_temperature = 52
print("Updated tool temperature:", tool_temperature, "degC")
````

````python
# Intermediate Example 2: Assignment using an expression

length = 8.5
width = 4.0

area = length * width
perimeter = 2 * (length + width)

print(f"Area = {area:.2f} m^2")
print(f"Perimeter = {perimeter:.2f} m")
````

## 2.5.3 Augmented Assignment Operators

Augmented assignment operators provide a shorter way to update the current value of a variable.

| Operator | Example | Equivalent statement |
|---|---|---|
| `+=` | `x += 5` | `x = x + 5` |
| `-=` | `x -= 2` | `x = x - 2` |
| `*=` | `x *= 3` | `x = x * 3` |
| `/=` | `x /= 2` | `x = x / 2` |
| `//=` | `x //= 2` | `x = x // 2` |
| `%=` | `x %= 3` | `x = x % 3` |
| `**=` | `x **= 2` | `x = x ** 2` |

````python
# Intermediate Example 3: Augmented assignment

score = 10
print("Initial score:", score)

score += 5
print("After score += 5:", score)

score -= 3
print("After score -= 3:", score)

score *= 2
print("After score *= 2:", score)

score /= 4
print("After score /= 4:", score)
````

## 2.5.4 Advanced Assignment Techniques

Python also supports assigning values to multiple variables in one statement and swapping values without using a temporary variable.

````python
# Advanced Example 1: Multiple assignment

length, width, height = 5.0, 3.0, 2.0

volume = length * width * height

print("Length:", length)
print("Width:", width)
print("Height:", height)
print("Volume:", volume)
````

````python
# Advanced Example 2: Assigning the same value to several variables

sensor_1 = sensor_2 = sensor_3 = 0

print(sensor_1, sensor_2, sensor_3)
````

````python
# Advanced Example 3: Swapping variable values

motor_a_speed = 1200
motor_b_speed = 1500

print("Before swapping:")
print("Motor A:", motor_a_speed, "rpm")
print("Motor B:", motor_b_speed, "rpm")

motor_a_speed, motor_b_speed = motor_b_speed, motor_a_speed

print("\nAfter swapping:")
print("Motor A:", motor_a_speed, "rpm")
print("Motor B:", motor_b_speed, "rpm")
````
