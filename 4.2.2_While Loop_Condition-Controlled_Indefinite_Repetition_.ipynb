## 4.2.2 `while` Loop — Condition-Controlled (Indefinite) Repetition

A `while` loop repeats statements **as long as a Boolean condition is `True`**. It is commonly used when the number of repetitions is **not known in advance**.

**Learning outcomes**
- Identify the condition and the repeated statements in a `while` loop.
- Write a `while` loop with correct indentation and a changing condition.
- Apply `while` loops to simple daily-life and mechanical engineering situations.

### How to write a `while` loop

```python
i = 1
while i <= 5:
    print(i)
    i = i + 1
```

- `while` is the keyword.
- `i <= 5` is the Boolean condition.
- The indented statements repeat while the condition is `True`.
- `i = i + 1` changes the value so the loop can stop.

**Important:** If the condition never becomes `False`, an infinite loop may occur.

### `for` versus `while`

| `for` loop | `while` loop |
|---|---|
| Commonly used when repetition count is known | Commonly used when repetition depends on a condition |
| `for i in range(5):` | `while i <= 5:` |

Both loops can repeat the same task. The choice depends on how the repetition is controlled.

### Pseudocode and flow of a `while` loop

```
START
    SET i = 1
    WHILE i <= 5
        OUTPUT i
        SET i = i + 1
    END WHILE
END
```

**Flow:** Start → Check condition → If `True`, execute statements and update → Check again → If `False`, exit.

### Using mathematical functions in Python

| Python code | Description |
|---|---|
| `import math` | Imports the math module |
| `math.radians(angle)` | Converts degrees to radians |
| `math.sin(angle)` | Calculates the sine of an angle in radians |
| `round(value, 2)` | Rounds a value to two decimal places |

```python
import math
angle = 30
angle_radian = math.radians(angle)
result = math.sin(angle_radian)
print("Sine value:", round(result, 2))
```

The sine function expects an angle in **radians**, not degrees.

### Example 1: Display Numbers from 1 to 5

**Case / Question**

A production line labels five components in sequence. Use a `while` loop to display the component numbers from **1 to 5**, stopping after the fifth component.

**Worked Solution**


````python
i = 1
while i <= 5:
    print(i)
    i = i + 1
````

### Example 2: Enter a Correct Password

**Case / Question**

A restricted machine control panel requires password **1234**. Ask the operator for a password and keep asking while it is incorrect. Display **Correct password!** when access is granted.

**Worked Solution**


````python
password = input("Enter password: ")
while password != "1234":
    print("Wrong password!")
    password = input("Enter password: ")
print("Correct password!")
````

### Example 3: Enter a Positive Number

**Case / Question**

A calculation requires a **positive integer**. Ask the user for a number and repeat the request while the number is zero or negative. Display the valid number.

**Worked Solution**


````python
num = int(input("Enter a number: "))
while num <= 0:
    print("Invalid number!")
    num = int(input("Enter a number: "))
print("Valid number:", num)
````

### Example 4: Countdown from 5 to 1

**Case / Question**

A machine starts a five-second countdown before operation. Display the countdown from **5 to 1**, then display **Finished!**.

**Worked Solution**


````python
i = 5
while i >= 1:
    print(i)
    i = i - 1
print("Finished!")
````

### Example 5: Calculate Total Cost Until the User Enters 0

**Case / Question**

A cashier enters item prices one at a time. Keep adding prices while the entered price is not **0**. When 0 is entered, display the total cost rounded to **two decimal places**.

**Worked Solution**


````python
total = 0
price = float(input("Enter item price (RM), 0 to stop: "))
while price != 0:
    total = total + price
    price = float(input("Enter item price (RM), 0 to stop: "))
print("Total cost: RM", round(total, 2))
````

### Example 6: Hydraulic Pressure Build-Up

**Case / Question**

A hydraulic pump starts at **0 bar** and increases pressure by **15 bar per stroke**. Display the pressure after every stroke until it reaches or exceeds **200 bar**, then display the total number of strokes.

**Worked Solution**


````python
pressure = 0
strokes = 0
while pressure < 200:
    pressure = pressure + 15
    strokes = strokes + 1
    print("Stroke", strokes, ":", pressure, "bar")
print("Total strokes:", strokes)
````

### Example 7: Cooling a Steel Component

**Case / Question**

A steel component initially has a temperature of **900°C** and retains **92% of its temperature each minute**. Display its temperature after each minute (one decimal place) until it falls below **100°C**, then display the cooling time.

**Worked Solution**


````python
temperature = 900.0
minute = 0
while temperature >= 100:
    temperature = temperature * 0.92
    minute = minute + 1
    print("Minute", minute, ":", round(temperature, 1), "C")
print("Cooling time:", minute, "minutes")
````
