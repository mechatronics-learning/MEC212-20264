## 4.2.2 `while` Loop — Condition-Controlled (Indefinite) Repetition
A `while` loop repeats statements **as long as a Boolean condition is `True`**. It is commonly used when the number of repetitions is **not known in advance**.

**Learning outcomes**
- Identify the condition and the repeated statements in a `while` loop.
- Write a `while` loop with correct indentation and a changing condition.
- Apply `while` loops to simple daily-life and mechanical engineering situations.


### How to write a `while` loop
```python
i = 1
while i <= 5:
    print(i)
    i = i + 1
```
- `while` is the keyword.
- `i <= 5` is the Boolean condition.
- The indented statements repeat while the condition is `True`.
- `i = i + 1` changes the value so the loop can stop.

**Important:** If the condition never becomes `False`, an infinite loop may occur.


### `for` versus `while`
| `for` loop | `while` loop |
|---|---|
| Commonly used when repetition count is known | Commonly used when repetition depends on a condition |
| Example: `for i in range(5):` | Example: `while i <= 5:` |

Both loops can repeat the same task. The choice depends on how the repetition is controlled.


### Pseudocode and flow of a `while` loop
```text
START
    SET i = 1
    WHILE i <= 5
        OUTPUT i
        SET i = i + 1
    END WHILE
END
```
**Flow:** Start → Check condition → If `True`, execute statements and update → Check again → If `False`, exit.

### Example 1: Display Numbers from 1 to 5


````python
i = 1
while i <= 5:
    print(i)
    i = i + 1
````

### Example 2: Enter a Correct Password


````python
password = input("Enter password: ")
while password != "1234":
    print("Wrong password!")
    password = input("Enter password: ")
print("Correct password!")
````

### Example 3: Enter a Positive Number


````python
num = int(input("Enter a number: "))
while num <= 0:
    print("Invalid number!")
    num = int(input("Enter a number: "))
print("Valid number:", num)
````

### Example 4: Countdown from 5 to 1


````python
i = 5
while i >= 1:
    print(i)
    i = i - 1
print("Finished!")
````

### Example 5: Calculate Total Cost Until the User Enters 0


````python
total = 0
price = float(input("Enter item price (RM), 0 to stop: "))
while price != 0:
    total = total + price
    price = float(input("Enter item price (RM), 0 to stop: "))
print("Total cost: RM", round(total, 2))
````

### Example 6: Hydraulic Pressure Build-Up


````python
pressure = 0
strokes = 0
while pressure < 200:
    pressure = pressure + 15
    strokes = strokes + 1
    print("Stroke", strokes, ":", pressure, "bar")
print("Total strokes:", strokes)
````

### Example 7: Cooling a Steel Component


````python
temperature = 900.0
minute = 0
while temperature >= 100:
    temperature = temperature * 0.92
    minute = minute + 1
    print("Minute", minute, ":", round(temperature, 1), "C")
print("Cooling time:", minute, "minutes")
````
