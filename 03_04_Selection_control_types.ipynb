## 3.4 One-Way Selection (`if`)

A **one-way selection** runs a block of code only when the condition is `True`. If the condition is `False`, the block is skipped entirely — nothing happens.

```python
if condition:        # colon required
    statement(s)    # indented 4 spaces — runs only when True
# program always continues here
```

Use a one-way `if` whenever you need to act on a single special case (e.g. trigger a warning) and do nothing otherwise.

````python
# Worked example — low coolant warning

coolant_level_pct = 18.0
LOW_COOLANT_PCT   = 20.0

if coolant_level_pct < LOW_COOLANT_PCT:
    print("Warning: coolant level low — refill before next shift")

print("Engine inspection complete")   # always runs
````

## 3.4 Two-Way Selection (`if–else`)

A **two-way selection** always takes exactly one of two paths. The `if` block runs when the condition is `True`; the `else` block runs when it is `False`. Unlike the one-way `if`, *something always happens*.

```python
if condition:
    statement(s)   # runs when True
else:
    statement(s)   # runs when False
```

Use `if–else` for binary engineering decisions: pass/fail, safe/unsafe, on/off.

````python
# Worked example — shaft stress pass/fail check

applied_stress_mpa = 310   # MPa — from load calculation
yield_strength_mpa = 250   # MPa — steel S275 yield strength

if applied_stress_mpa > yield_strength_mpa:
    print("FAIL: applied stress exceeds yield strength — redesign required")
else:
    print("PASS: design is safe")
````

## 3.4 Multi-Way Selection (`if–elif–else`)

A **multi-way selection** tests a series of conditions in order and executes the block belonging to the **first** condition that is `True`. All remaining conditions are then skipped. The optional `else` catches any case that did not match.

```python
if condition1:
    statement(s)    # tested first
elif condition2:
    statement(s)    # tested only if condition1 is False
elif condition3:
    statement(s)
else:
    statement(s)    # runs if nothing above matched
```

> Only the **first** matching branch runs — order conditions from most critical to least critical.

````python
# Worked example — motor RPM classification

rpm = 2200   # try: -100, 0, 500, 2200, 4000

if rpm < 0:
    print("Error: invalid RPM — cannot be negative")
elif rpm == 0:
    print("Motor stopped")
elif rpm <= 1000:
    print("Low-speed mode")
elif rpm <= 3000:
    print("Normal operating speed")
else:
    print("Warning: overspeed — check drive settings")
````

## 3.5 Nested `if` Statements

A **nested `if`** is an `if` statement placed *inside* another `if` (or `elif`/`else`) block. The inner `if` is only reached when the outer condition is `True`, creating a two-level (or deeper) decision tree.

```python
if outer_condition:          # Level 1
    if inner_condition:      # Level 2 — only reached when outer is True
        statement(s)
    else:
        statement(s)
else:
    statement(s)             # Level 1 False branch
```

Each level adds 4 spaces of indentation. More than 3 levels usually signals the logic should be restructured using `and`/`or`.

````python
# Worked example — pressure vessel safety (temperature and pressure combined)

temp_c   = 180.0
pres_bar = 35.0

if temp_c <= 200:                              # outer: temperature acceptable?
    if pres_bar <= 50:                         # inner: pressure also OK?
        print("Safe operating condition")
    else:
        print("Warning: high pressure at normal temperature")
else:                                          # outer False: temp too high
    if pres_bar <= 30:                         # reduced pressure limit
        print("Warning: high temperature — reduced pressure limit active")
    else:
        print("DANGER: both temperature and pressure exceed limits")
````
