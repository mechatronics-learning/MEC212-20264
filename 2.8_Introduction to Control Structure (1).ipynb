# 2.8 Introduction to Type of Control Structure

A **control structure** determines the order in which program statements are executed.

Control structures allow a program to:

- Execute statements sequentially
- Make decisions
- Repeat a set of instructions

There are three primary types of control structures in Python:

1. Sequential Control (Implicit)
2. Selection Control (Conditional Statements)
3. Iteration Control (Looping Statements)


## 2.8.1 Sequential Control (Implicit)

Sequential control is the most basic control structure.

Instructions are executed one after another, in the order they appear in the program, from top to bottom. This is the default behaviour when no other control structures are used.


````python
print("Step 1: Start")

x = 10
y = 20
z = x + y

print(f"Step 2: Sum is {z}")
print("Step 3: End")
````

### Expected Output

```text
Step 1: Start
Step 2: Sum is 30
Step 3: End
```


## 2.8.2 Selection Control (Conditional Statements)

Selection structures allow a program to make decisions and execute different blocks of code based on whether a condition is `True` or `False`.

Python uses:

- `if`
- `elif`
- `else`

for this purpose.


````python
temperature = 28

if temperature > 25:
    print("It's hot outside!")
else:
    print("It's not too hot.")
````

## 2.8.3 Iteration Control (Looping Statements)

Iteration structures, also known as loops, allow a block of code to be executed repeatedly.

Python commonly uses:

- `for` loop
- `while` loop


### Example 1: `for` loop

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

### Example 2: `range()` with a `for` loop

````python
for i in range(5):
    print(f"Count: {i}")
````
