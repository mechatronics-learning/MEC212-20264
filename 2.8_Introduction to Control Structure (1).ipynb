## 2.8 Introduction to Types of Control Structures


A **control structure** determines the order in which program statements are executed. Control structures allow a program to:

- Execute statements sequentially
- Make decisions
- Repeat a set of instructions

There are **three main types** of control structures in Python:

1. **Sequential Control (Implicit)**
2. **Selection Control (Conditional Statements)**
3. **Iteration Control (Looping Statements)**

## 2.8.1 Sequential Control (Implicit)

Sequential control is the most basic control structure. Instructions are executed one after another, **from top to bottom**, in the order they appear in the program. This is the default behaviour when no other control structures are used.

### Example: Adding Two Numbers

````python
print("Step 1: Start")
x = 10
y = 20
z = x + y
print("Step 2: Sum is", z)
print("Step 3: End")
````

**Expected Output**
```text
Step 1: Start
Step 2: Sum is 30
Step 3: End
```

## 2.8.2 Selection Control (Conditional Statements)

Selection control allows a program to **make decisions** and execute different instructions depending on whether a condition is `True` or `False`.

Python uses `if`, `elif` and `else` for selection. You only need to recognise the purpose of these keywords at this stage.

### Example: Temperature Check

````python
temperature = 28

if temperature > 25:
    print("It's hot outside!")
else:
    print("It's not too hot.")
````

**Expected Output**
```text
It's hot outside!
```
The condition `temperature > 25` is true, so the first message is displayed.

## 2.8.3 Iteration Control (Looping Statements)

Iteration control, also called **looping**, allows a set of instructions to be executed repeatedly. Python commonly uses `for` and `while` loops. You only need to understand that loops repeat instructions at this stage.

### Example 1: Repeating Items in a List

````python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
````

**Expected Output**
```text
apple
banana
cherry
```

### Example 2: Repeating with `range()`

````python
for i in range(5):
    print("Count:", i)
````

**Expected Output**
```text
Count: 0
Count: 1
Count: 2
Count: 3
Count: 4
```
`range(5)` produces five numbers, starting from 0.
