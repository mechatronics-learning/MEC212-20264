## 4.3 Nested Loops

A **nested loop** is a loop placed inside another loop. The **inner loop completes all its iterations** for each single iteration of the outer loop:

- Total iterations = outer count × inner count
- The outer loop typically controls rows / major steps
- The inner loop typically controls columns / minor steps

Nested loops are commonly used in engineering to process **two-dimensional data** such as sensor grids, matrices, or parameter sweeps.


````python
# Worked example — average temperature across a 3×4 sensor grid (°C)
grid = [
    [72.1, 74.3, 75.0, 73.8],   # row 0
    [80.2, 81.5, 79.9, 82.0],   # row 1
    [68.4, 67.9, 70.1, 69.5],   # row 2
]

total = 0.0
count = 0

for row_index, row in enumerate(grid):      # outer: each row
    for col_index, temp in enumerate(row):  # inner: each sensor in the row
        total += temp
        count += 1

print(f"Total readings : {count}")
print(f"Grid average   : {total / count:.2f} °C")
````
