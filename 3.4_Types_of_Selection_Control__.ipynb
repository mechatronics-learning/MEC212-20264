## Recall: Boolean and relational/logical operators

A Boolean condition evaluates to **`True` or `False`**. Relational operators (`>`, `<`, `>=`, `<=`, `==`, `!=`) compare values; logical operators (`and`, `or`, `not`) combine or reverse conditions. **Chapter 3.3 builds conditions; Chapter 3.4 uses those conditions to choose actions.**

**Python rules:** use a colon `:` after `if`, `elif`, `else`, `match` and `case` headers; indent code inside each branch; use `==` for comparison, not `=`. `input()` returns text, so convert it using `int()` or `float()` when needed. **Nested `if` belongs to Chapter 3.5**, not this tutorial.


````python
temperature = 95
pressure = 140
print(temperature > 85)
print((temperature > 85) and (pressure > 120))
````

## One-way selection: `if`

Run an action **only when** a condition is `True`. If it is `False`, skip the indented block.

```python
if condition:
    # action
```


## Two-way selection: `if...else`

Choose **one of two** paths. `else` has no condition.

```python
if condition:
    # action A
else:
    # action B
```


## Multi-way selection: `if...elif...else`

Check conditions **top to bottom**. The first `True` branch runs; `else` is the fallback.

```python
if condition1:
    # action A
elif condition2:
    # action B
else:
    # default
```


## Multi-way selection: `match...case`

Match **one value** against several exact cases. `case _` is the default. Requires Python 3.10+.

```python
match value:
    case 1:
        # action A
    case 2:
        # action B
    case _:
        # default
```

### **Example 1 — Motor Overspeed Warning (`if`)**

A motor rotates at 1850 rpm. Display a warning only if speed exceeds 1800 rpm.


````python
speed_rpm = 1850
if speed_rpm > 1800:
    print("Warning: Motor overspeed")
print("Monitoring complete")
````

**Expected output**

```text
Warning: Motor overspeed
Monitoring complete
```


### **Example 2 — Hydraulic Pressure Status (`if...else`)**

A hydraulic system is acceptable when pressure is at least 120 kPa.


````python
pressure_kpa = 135
if pressure_kpa >= 120:
    print("Pressure acceptable")
else:
    print("Pressure too low")
````

**Expected output**

```text
Pressure acceptable
```


### **Example 3 — Vibration Severity (`if...elif...else`)**

Classify vibration amplitude: at least 8 mm/s = Critical; at least 4 mm/s = Warning; otherwise = Normal.


````python
vibration = 5.5
if vibration >= 8:
    print("Critical")
elif vibration >= 4:
    print("Warning")
else:
    print("Normal")
````

**Expected output**

```text
Warning
```

## Quick review

- `if`: execute an action only if the condition is true.

- `match...case`: choose based on a matching value (Python 3.10+).
