## 1. Connection to Chapter 3.2 — Boolean Values

A Boolean is either **`True`** or **`False`**. A Boolean variable stores a value, whereas a **relational expression calculates a Boolean result by comparing values**. A **logical expression combines or reverses Boolean conditions**.

`temperature_ok = True` → Boolean value; `temperature < 90` → relational expression; `(temperature < 90) and (pressure < 200)` → logical expression.

In **Chapter 3.4**, these conditions will be used in `if` statements to select actions.

## 2. Relational Operators

| Operator | Meaning | Example | Result |
|:---:|---|---|:---:|
| `>` | greater than | `12 > 10` | `True` |
| `<` | less than | `12 < 10` | `False` |
| `>=` | greater than or equal | `12 >= 12` | `True` |
| `<=` | less than or equal | `12 <= 10` | `False` |
| `==` | equal to | `12 == 12` | `True` |
| `!=` | not equal to | `12 != 10` | `True` |

**Important:** `=` assigns a value; `==` compares two values. Strings are case-sensitive: `"ON" == "on"` is `False`.

## 3. Logical Operators

- **`and`** → `True` only when **both** conditions are `True`.
- **`or`** → `True` when **at least one** condition is `True`.
- **`not`** → reverses `True` ↔ `False`.

| A | B | A and B | A or B |
|:---:|:---:|:---:|:---:|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

Use parentheses for readability: `(speed <= 100) and (temperature < 80)`.

**Precedence:** comparisons are evaluated before `not`, then `and`, then `or`. Parentheses make the intended logic clearer. The output is Boolean, **not** a machine command.

## 4. Worked Examples (Run Each Cell)

The examples below use `print()` to show every result. The exercises later ask you to write and evaluate Boolean expressions without `if`/`else`.

### Example 1 — Hydraulic Cylinder Stroke

```python
stroke_mm = 145
maximum_stroke_mm = 150
print(stroke_mm <= maximum_stroke_mm)  # True
```

### Example 2 — Shaft Rotation Direction

```python
rotation_direction = "CW"
print(rotation_direction == "CW")   # True
print(rotation_direction != "CCW")  # True
```

### Example 3 — Cooling System Interlock

```python
coolant_flow_ok = True
fan_running = False
print(coolant_flow_ok and fan_running)  # False
print(coolant_flow_ok or fan_running)   # True
```

### Example 4 — Gearbox Monitoring

```python
oil_temperature_c = 76
vibration_mm_s = 3.2
print((oil_temperature_c < 85) and (vibration_mm_s <= 4.5))  # True
```

### Example 5 — Pressure Alarm Condition

```python
line_pressure_kpa = 260
pressure_limit_kpa = 240
sensor_fault = False
print((line_pressure_kpa > pressure_limit_kpa) or sensor_fault)  # True
print(not sensor_fault)  # True
```
