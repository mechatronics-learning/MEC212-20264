## 1. Notes — What is a nested `if`?

A **nested `if`** is an `if` statement inside another `if` or `else` block. The inner decision is reached only when execution enters its enclosing block.

**Think:** First check a general requirement; only then check a more specific requirement.

````
if first_condition:
    if second_condition:
        print("Both conditions satisfied")
    else:
        print("First passed, second failed")
else:
    print("First condition failed")

````

**Indentation:** Use four spaces for each level. The inner `else` belongs to the inner `if`; the outer `else` belongs to the outer `if`.

**Without `else`:** It is valid to write nested `if` statements without any `else`. If a condition is False, the nested block is skipped and the program may display nothing.

**Connection to Chapter 3.3:** Comparisons (`>=`, `<`, `==`) evaluate to `True` or `False`. Logical operators (`and`, `or`, `not`) can combine or reverse these conditions. A nested `if` is useful when you need different actions at different decision stages.

**Nested `if` vs `if...elif...else`:** Nested `if` represents *dependent checks*. `if...elif...else` chooses between *alternative outcomes*. For a simple yes/no result, `if A and B:` can sometimes replace two nested checks, but nesting is clearer when the failure reasons must be different.

## 2. Three worked examples

Each example uses a different mechanical engineering application. Run the code, then change the input values to explore each branch.

### Example 1 — CNC Machine Door Interlock

**Sample test:** Power ON = yes; Door closed = no → Close the safety door.

`````python
# Example 1: CNC Machine Door Interlock
power_on = input("Power ON? (yes/no): ")
door_closed = input("Door closed? (yes/no): ")
if power_on == "yes":
    if door_closed == "yes":
        print("CNC machine ready")
    else:
        print("Close the safety door")
else:
    print("Switch on the machine first")
`````

### Example 2 — Bearing Lubrication Inspection (without else)

**Sample test:** Due = yes; Grease available = yes → Perform bearing lubrication. Other inputs produce no message.

`````python
# Example 2: Bearing Lubrication Inspection (without else)
lubrication_due = input("Lubrication due? (yes/no): ")
grease_available = input("Grease available? (yes/no): ")
if lubrication_due == "yes":
    if grease_available == "yes":
        print("Perform bearing lubrication")
`````

### Example 3 — Cooling Fan Control (three levels)

**Sample test:** Running = yes; Coolant = 102; Fan functional = no → Warning: cooling fan fault.

`````python
# Example 3: Cooling Fan Control (three levels)
engine_running = input("Engine running? (yes/no): ")
coolant_c = float(input("Coolant temperature (C): "))
fan_functional = input("Cooling fan functional? (yes/no): ")
if engine_running == "yes":
    if coolant_c > 95:
        if fan_functional == "yes":
            print("Activate cooling fan")
        else:
            print("Warning: cooling fan fault")
    else:
        print("Cooling fan not required")
else:
    print("Engine is stopped")
`````
