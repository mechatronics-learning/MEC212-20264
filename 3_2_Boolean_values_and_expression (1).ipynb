## Boolean values

A Boolean (`bool`) has exactly two values: `True` and `False`. Python is case-sensitive. Use `type()` to inspect a data type.


## Boolean expressions

An expression such as `force_n > 400` evaluates to `True` or `False`. Relational operators are covered in detail in 3.3.


## Boolean conversion

`bool(0)` and `bool("")` are `False`. Nonzero numbers and nonempty strings are `True`. `int(True)` is 1 and `int(False)` is 0.


### Notes demonstration

```python
print(type(True))
print(type(False))
print(bool(0))
print(bool(25))
print(bool(""))
print(bool("RUN"))
print(int(True))
print(int(False))
```
