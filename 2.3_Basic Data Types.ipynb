## Solution 2.3.1

````python
age = 20
height = 1.75
student_name = "Aina"
is_registered = True

print(type(age))
print(type(height))
print(type(student_name))
print(type(is_registered))
````

### Expected Output

```text
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
```

## Solution 2.3.2

````python
value1 = 250
value2 = 36.8
value3 = "Mechanical Engineering"
value4 = False
value5 = [12, 25, 30]
value6 = {"Temperature": 36.5}

print(type(value1))
print(type(value2))
print(type(value3))
print(type(value4))
print(type(value5))
print(type(value6))
````

### Expected Output

```text
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
<class 'list'>
<class 'dict'>
```

## Solution 2.3.3

````python
machine_id = "MC-102"
temperature = 68.5
running_status = True
number_of_products = 150

print("Machine ID:", machine_id, "-", type(machine_id))
print("Temperature:", temperature, "-", type(temperature))
print("Running Status:", running_status, "-", type(running_status))
print("Number of Products:", number_of_products, "-", type(number_of_products))
````

## Solution 2.3.4

````python
student = {
    "Name": "Aina",
    "Age": 20,
    "CGPA": 3.82,
    "Registered": True
}

print("Student Information:", student)
print("Dictionary Type:", type(student))

print("Name Type:", type(student["Name"]))
print("Age Type:", type(student["Age"]))
print("CGPA Type:", type(student["CGPA"]))
print("Registered Type:", type(student["Registered"]))
````
