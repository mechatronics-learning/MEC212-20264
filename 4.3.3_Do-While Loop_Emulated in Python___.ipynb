## 1. Introduction
A **do-while loop** executes its statements **at least once**, then checks whether to repeat.

**Concept (C/C++):**
```cpp
do {
    // statements
} while (condition);
```

**Python equivalent:**
```python
while True:
    # statements
    if not condition:
        break
```

The two examples use opposite tests: the conceptual `do-while` **continues** when its condition is true; the Python `if` checks when to **exit**.

### Key characteristics of `while True` and `break`
- `while True` begins a loop that keeps running until it is stopped.
- Statements inside the loop run at least once.
- An `if` statement checks an exit condition.
- `break` immediately exits the loop when that condition is met.
- If `break` is not executed, the loop repeats.

**Remember:** `break` exits the loop; `continue` skips to the next iteration. This tutorial focuses on `break`.

## 2. Worked Examples
Run each example in a separate Google Colab code cell. When a cell uses `input()`, type the requested values.

### Example 1 — Enter a Positive Number

**Python Code**

````python
while True:
    number = int(input("Enter a positive number: "))
    if number > 0:
        break
    print("Invalid number. Try again.")

print("Valid number:", number)
````

### Example 2 — Enter a Positive Pressure

**Python Code**

````python
while True:
    pressure = float(input("Enter pressure (bar): "))
    if pressure > 0:
        break
    print("Invalid pressure. Try again.")

print("Pressure accepted:", pressure, "bar")
````

### Example 3 — Enter a Correct Password

**Python Code**

````python
while True:
    password = input("Enter password: ")
    if password == "1234":
        break
    print("Wrong password. Try again.")

print("Access granted!")
````

### Example 4 — Enter a Valid Temperature

**Python Code**

````python
while True:
    temperature = float(input("Enter temperature (0 to 100 C): "))
    if temperature >= 0 and temperature <= 100:
        break
    print("Temperature outside the range.")

print("Accepted temperature:", temperature, "C")
````

### Example 5 — Choose a Menu Option

**Python Code**

````python
while True:
    choice = int(input("Choose 1 (Start) or 2 (Stop): "))
    if choice == 1 or choice == 2:
        break
    print("Invalid choice.")

if choice == 1:
    print("System started")
else:
    print("System stopped")
````
