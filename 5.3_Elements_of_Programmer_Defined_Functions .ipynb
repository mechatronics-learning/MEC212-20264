# 5.3 Elements of Programmer-Defined Functions

This notebook covers the core elements of programmer-defined functions and includes optional enrichment material from the lecture slides.

## Functions with No Parameters

A function can take zero parameters.

````python
def get_pi():
    """Return the value of pi (approximate)."""
    return 3.14159

result = get_pi()
print(result)
````

### Mechanical Engineering Example — No Parameters

````python
def get_gravity():
    """Return standard gravitational acceleration."""
    return 9.81

g = get_gravity()
print("Gravity =", g, "m/s²")
````

## Functions with No Return Value

A function can perform an action without returning a value.

````python
def display_report(force, area):
    """Display a formatted stress report."""
    stress = force / area
    print(f"Force: {force} N")
    print(f"Area: {area} m²")
    print(f"Stress: {stress:.2f} Pa")
    # no return statement

display_report(500, 0.05)
````

## Elements of a Function

The **signature** and **body** together define a function.

### Signature (interface)
- Function name
- Parameter list

### Body (implementation)
- Local variables
- Statements / algorithm
- `return` statement (if returning a value)

## Optional Enrichment — Parameter Styles

These are useful Python features but are not essential for the core MEC212 function syllabus.

### Positional Arguments

````python
def beam_bending(length, load):
    deflection = (load * length**3) / (3 * 29e6)
    return deflection

d = beam_bending(2.0, 1000)
print(d)
````

### Keyword / Default Parameters

````python
def material_properties(name, E=200, v=0.3):
    return E, v

E1, v1 = material_properties("Steel")
E2, v2 = material_properties("Aluminum", E=70)

print(E1, v1)
print(E2, v2)
````

## Optional Enrichment — Default Parameter Values

````python
def convert_temperature(celsius, output="fahrenheit"):
    if output == "fahrenheit":
        return (celsius * 9/5) + 32
    elif output == "kelvin":
        return celsius + 273.15

t1 = convert_temperature(25)
t2 = convert_temperature(25, "kelvin")

print(t1)
print(t2)
````

## Return Values

A function can return one value, multiple values, or no value.

### Single Return

````python
def area_circle(radius):
    return 3.14159 * radius**2

print(area_circle(5))
````

### Optional Enrichment — Multiple Returns

````python
def circle_properties(radius):
    """Return area and circumference."""
    area = 3.14159 * radius**2
    circumference = 2 * 3.14159 * radius
    return area, circumference

a, c = circle_properties(5)
print("Area =", a)
print("Circumference =", c)
````

### No Return

````python
def log_event(message):
    print(f"[LOG] {message}")
    # no return -> Python returns None implicitly

result = log_event("Machine started")
print("Returned value:", result)
````

## Optional Enrichment — Local Variables and Scope

**Scope** is the region of a program where a variable is defined and accessible.

Variables created inside a function are **local variables**.

````python
def calculate():
    x = 10
    y = 20
    return x + y

result = calculate()
print(result)

# Uncomment the next line to observe the error:
# print(x)
````

A variable created outside functions is a **global variable**. For beginner programming, it is better to pass data using parameters and return results instead of relying on globals.

## Optional Engineering Design Example — Safety Factor

This example also uses selection (`if/else`) and error handling, so it is better treated as enrichment after students are comfortable with functions.

````python
def safety_factor(applied_stress, yield_strength):
    if applied_stress <= 0:
        raise ValueError("Applied stress must be positive")
    return yield_strength / applied_stress

yield_steel = 250
applied = 200
sf = safety_factor(applied, yield_steel)

print(f"Safety factor: {sf:.2f}")

if sf >= 1.5:
    print("Design is acceptable")
else:
    print("Design needs improvement")
````

## Predefined Functions vs Programmer-Defined Functions

### Predefined functions
Examples:
- `print()`, `len()`, `range()`, `int()`, `float()`, `abs()`, `max()`, `min()`
- `math.sqrt()`, `math.sin()`, `math.cos()`

### Programmer-defined functions
Created for a specific problem, for example:
- `calculate_stress()`
- `calculate_strain()`
- `safety_factor()`

## Common Syntax Pitfalls

| Mistake | Incorrect | Correct |
|---|---|---|
| Forgetting colon | `def func(x)` | `def func(x):` |
| Incorrect indentation | body at column 0 | body indented |
| Missing return when a value is needed | compute only | `return value` |
| Wrong number of arguments | `func(1)` when 2 required | `func(1, 2)` |
| Forgetting parentheses when calling | `result = calculate` | `result = calculate()` |

## Chapter Summary

| Concept | Definition | Example |
|---|---|---|
| Function | Reusable block performing one task | `def add(a, b): return a + b` |
| Parameter | Variable in function definition | `a, b` |
| Argument | Value passed when calling | `5, 3` |
| Return value | Value sent back to caller | `8` |
| Local variable | Variable created inside a function | `x` inside `def f()` |
