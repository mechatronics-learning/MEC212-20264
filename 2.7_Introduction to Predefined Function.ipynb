# 2.7 Introduction to Predefined Functions

## What is a Predefined Function?

Predefined functions are functions that are already built into Python.
They are also known as **built-in functions** because they are provided by Python and can be used directly without creating your own function.

A predefined function performs a specific task simply by calling its name.


## Commonly Used Predefined Functions

| Function | Purpose | Example |
|---|---|---|
| `print()` | Display output | `print("Hello")` |
| `input()` | Receive user input | `name = input()` |
| `len()` | Count number of characters/items | `len("Python")` |
| `type()` | Identify data type | `type(3.14)` |
| `int()` | Convert to integer | `int("25")` |
| `float()` | Convert to float | `float("3.5")` |
| `str()` | Convert to string | `str(25)` |
| `round()` | Round a number | `round(3.567,2)` |
| `max()` | Largest value | `max(5,8,2)` |
| `min()` | Smallest value | `min(5,8,2)` |
| `abs()` | Absolute value | `abs(-15)` |
| `sum()` | Sum values | `sum([1,2,3])` |


## Example 1: `print()`

```python
print("print() is a predefined function")
```

## Example 2: `len()`

```python
countCharacter = len("thermodynamics")
print(countCharacter)
```

## Example 3: `type()`

```python
findType = type(3.142)
print(findType)
```

## Example 4: Type Conversion

```python
age = "20"
age = int(age)
print(age)
print(type(age))
```
