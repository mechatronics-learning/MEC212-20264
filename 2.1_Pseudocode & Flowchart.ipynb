**Exercise 2.1 – Pseudocode**

Question - Write a pseudocode to calculate the area of a rectangle.

The program should:

Read the length.
Read the width.
Calculate the area.
Display the area.
Task

Write the complete pseudocode.

```python
# BEGIN
#     INPUT length
#     INPUT width
#     SET area ← length * width
#     DISPLAY area
# END
```

```python
#Translating Pseudocode into Python
length = float(input("Enter length: "))
width = float(input("Enter width: "))
area = length * width
print("Area =", area)
```

**Exercise 2.2 – Pseudocode**

Question- Write a pseudocode to calculate the average mark of three subjects.

Write the complete pseudocode.

```python
# BEGIN
#     INPUT mark1
#     INPUT mark2
#     INPUT mark3
#
#     SET total ← mark1 + mark2 + mark3
#     SET average ← total / 3
#
#     DISPLAY total
#     DISPLAY average
# END
```

```python
# Translating pseudocode in Python
# Exercise 2.5 – Calculate the Average Mark of Three Subjects

# Input three subject marks
mark1 = float(input("Enter Mark 1: "))
mark2 = float(input("Enter Mark 2: "))
mark3 = float(input("Enter Mark 3: "))

# Calculate the total and average
total = mark1 + mark2 + mark3
average = total / 3

# Display the results
print("\n----- Result -----")
print("Total Marks   =", total)
print("Average Marks =", average)
```

**Exercise 2.3  –Pseudocode**

Question- Write a pseudocode to calculate the Body Mass Index (BMI).

The formula is:

BMI = Weight (kg) / Height² (m²)

Task

Write the complete pseudocode.

```python
# BEGIN
#     INPUT weight
#     INPUT height
#
#     SET bmi ← weight / (height × height)
#
#     DISPLAY bmi
# END
```

```python
# Translating pseudocode in Python
# Input weight and height
weight = float(input("Enter your weight (kg): "))
height = float(input("Enter your height (m): "))

# Calculate BMI
bmi = weight / (height * height)

# Display the result
print("\n----- Result -----")
print(f"Your BMI = {bmi:.2f}")
```
